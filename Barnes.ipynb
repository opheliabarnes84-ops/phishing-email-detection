```python
import pandas as pd
import os

# Load dataset from kagglehub path. It is assumed that 'Phishing_Email.txt'
# is extracted into the 'path' directory by kagglehub.
df = pd.read_csv(os.path.join(path, '/content/archive.zip'))

# Dataset shape
print("Dataset Shape:")
print(df.shape)

# Column names
print("\nColumns:")
print(df.columns)

# Data types
print("\nData Types:")
print(df.dtypes)

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Class balance
print("\nClass Distribution:")
print(df['Email Type'].value_counts())

# Sample rows
print("\nFirst 5 Rows:")
print(df.head())
```
